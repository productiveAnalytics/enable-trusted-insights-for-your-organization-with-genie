# Lecture - Introduction to Genie Agents
#### Duration - 10 Minutes

## Overview

A **Genie Agent** is a curated, natural language interface to a domain's Unity Catalog data, enabling business users to interact with that data using plain English. It uses generative AI tailored to your organization's data and terminology, and it can be continuously improved through monitoring and user feedback.

Rather than requiring business users to ask data analysts for help, Genie Agents allow anyone to ask questions in plain English and receive SQL-backed answers, tables, and visualizations, all grounded in governed Unity Catalog data.

In this lecture, you will quickly review the Databricks AI/BI product family:

- AI/BI Dashboards
- Genie Agents
- Genie One *(formerly Databricks One)*
- Genie Code

You will learn how they complement each other, then focus on how a Genie Agent turns a natural-language question into a SQL result, and how Unity Catalog governance applies.

## Learning Objectives

By the end of this lecture, you will be able to:

1. **Describe the Databricks AI/BI product family** (AI/BI Dashboards, Genie Agents, Genie One, and Genie Code) and how they differ.
2. **Explain how a Genie Agent turns a natural-language question into a SQL-backed answer.**
3. **Describe how Unity Catalog governance and security apply** to the results a Genie Agent returns.

##### EXPAND FOR ADDITIONAL NOTES

<details>

This flow shows the core idea behind Genie Agents.

A user asks a question in plain, everyday language. No SQL, no dashboards to navigate, no need to know where the data lives.

A Genie Agent takes that question, understands the intent, and uses your governed data to generate a response.

That response can include queries, summaries, and visualizations.

### Consumers
- Can ask questions the same way you would ask a colleague
- A Genie Agent translates your question into data queries behind the scenes
- Results are grounded in your organization's data, not generic answers
- You can iterate, refine, and explore without needing technical skills

Think of it as moving from "I need to write SQL" to "I just ask what I want to know."

## C. Conclusion

In this lecture, you explored the Databricks AI/BI product family and the role of Genie Agents in enabling self-service data exploration. Key concepts covered include:

- **AI/BI product family** - Dashboards deliver curated, repeatable answers to known questions; Genie Agents let business users ask ad-hoc questions in natural language without writing SQL.
- **Genie One** - the simplified, full-screen chat experience that gives business users a single entry point to ask questions in plain English, without navigating compute, queries, models, or notebooks.
- **Question-to-answer flow** - a user asks a question, the Genie Agent interprets it using available context, generates SQL, runs it on a SQL warehouse, and returns a transparent, data-backed result.
- **Governed and secure** - Unity Catalog permissions and security controls apply to the data a Genie Agent uses and the results it returns.
