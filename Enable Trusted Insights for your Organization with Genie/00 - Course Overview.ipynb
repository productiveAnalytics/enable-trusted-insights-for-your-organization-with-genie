# Enable Trusted Insights for your Organization with Genie
 

## Overview

A **Genie Agent** is a curated, natural-language interface to your organization's Unity Catalog data. It lets business users ask questions in plain English and get SQL-backed answers, tables, and visualizations, all grounded in and governed by **Unity Catalog**.

In this hands-on workshop, you build a Genie Agent on the **bakehouse** dataset (a fictional bakery franchise) from the ground up. You create the agent, explore how it behaves without any curation, then curate its **Knowledge Store** with metadata, SQL logic, and instructions to make it reliable, using benchmarks to measure the improvement. Finally, you share the agent, experience it as an end user through **Genie One**, and use the **Monitoring** tab to turn real usage into your next round of curation.

By the end, you will have a working, curated Genie Agent and a repeatable workflow for delivering trusted, natural-language insights across your organization.

## Terminal Objectives

By the end of this workshop, you will be able to:

1. **Describe the Databricks AI/BI and Genie product family** and explain how a Genie Agent turns a natural-language question into a governed, SQL-backed answer.
2. **Create a Genie Agent** on Unity Catalog tables and navigate its interface.
3. **Curate a Genie Agent's Knowledge Store** with metadata, SQL logic, and instructions, and use benchmarks to measure the improvement.
4. **Share a Genie Agent, experience it in Genie One, and use monitoring** to drive continuous improvement.

## A. Prerequisites

Before starting this course, learners should be comfortable with the following:

- **General familiarity with the Databricks workspace** - You can navigate the workspace UI, attach compute resources, and run notebooks.

- **Familiarity with Unity Catalog concepts** - You understand catalogs, schemas, and tables in Unity Catalog. Prior experience with permissions, tags, or column masks is helpful but not required.

- **Intermediate SQL proficiency** - You can read and write SQL queries including SELECT, JOIN, GROUP BY, and aggregate functions.

- **Access to a Databricks workspace** with the Databricks SQL entitlement and CAN USE permission on a serverless SQL warehouse.
